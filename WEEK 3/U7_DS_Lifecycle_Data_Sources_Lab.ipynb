```python
import numpy as np
from scipy.stats import entropy
from sklearn.datasets import load_iris

# 1. Entropy of a fair 4-sided and 6-sided die (base=2)
die_4 = np.ones(4) / 4
die_6 = np.ones(6) / 6

entropy_4 = entropy(die_4, base=2)
entropy_6 = entropy(die_6, base=2)

print("Entropy of 4-sided die:", entropy_4)
print("Entropy of 6-sided die:", entropy_6)


# 2. KL divergence D(P || Q)
P = np.array([0.7, 0.3])
Q = np.array([0.5, 0.5])

kl_divergence = entropy(P, Q)

print("KL divergence D(P || Q):", kl_divergence)


# 3. Most / least informative Iris feature
iris = load_iris()

X = iris.data
y = iris.target
feature_names = iris.feature_names

feature_entropies = []

for i in range(X.shape[1]):
    values = X[:, i]

    # Calculate entropy of the feature values
    _, counts = np.unique(values, return_counts=True)
    probabilities = counts / len(values)

    feature_entropy = entropy(probabilities, base=2)
    feature_entropies.append(feature_entropy)

most_informative_index = np.argmax(feature_entropies)
least_informative_index = np.argmin(feature_entropies)

most_informative = feature_names[most_informative_index]
least_informative = feature_names[least_informative_index]

print("Most informative:", most_informative)
print("Least informative:", least_informative)

```

```python
mapping = {
    'Define':   'Business Understanding',
    'Collect':  'Data Understanding',
    'Explore':  'Data Understanding',
    'Prepare':  'Data Preparation',
    'Model':    'Modeling',
    'Evaluate': 'Evaluation',
    'Deploy':   'Deployment',
    'Monitor':  'Deployment',
}
for stage, phase in mapping.items():
    print(f'{stage:9s} -> {phase}')

```

```python
# Stage -> Phase mapping
stage_to_phase = {
    'Collect': 'Data',
    'Prepare': 'Data',
    'Explore': 'Data',
    'Model': 'Modeling',
    'Evaluate': 'Modeling',
    'Deploy': 'Deployment',
    'Monitor': 'Deployment',
    'Define': 'Planning'
}

# 1. Invert the mapping: group stages by phase
grouped = {}

for stage, phase in stage_to_phase.items():
    if phase not in grouped:
        grouped[phase] = []
    grouped[phase].append(stage)

# 2. Print phase -> stages
for phase, stages in grouped.items():
    print(phase, "->", stages)

# 3. Which phase has the most stages?
phase_with_most_stages = max(grouped, key=lambda phase: len(grouped[phase]))

print("Phase with most stages:", phase_with_most_stages)
```

```python
def classify(fmt):
    fmt = fmt.lower()
    if fmt in ('csv', 'sql', 'parquet', 'excel', 'xlsx'):
        return 'structured'
    if fmt in ('json', 'xml', 'yaml', 'log'):
        return 'semi-structured'
    return 'unstructured'   # text, image, audio, video, ...

for fmt in ['csv', 'json', 'jpg', 'sql', 'xml', 'mp3']:
    print(f'{fmt:6s} -> {classify(fmt)}')


```

```python
from collections import Counter

to_label = ['parquet', 'yaml', 'png', 'csv', 'wav', 'json', 'txt']

# 1. Classify each one
categories = {
    'parquet': 'structured',
    'yaml': 'semi-structured',
    'png': 'unstructured',
    'csv': 'structured',
    'wav': 'unstructured',
    'json': 'semi-structured',
    'txt': 'unstructured'
}

labels = []

for file_type in to_label:
    labels.append(categories[file_type])

print("Classifications:", dict(zip(to_label, labels)))

# 2. Count per category
category_counts = Counter(labels)

print("Count per category:", category_counts)

# 3. An ML task that uses unstructured data
ml_task = "Image classification"

print("An ML task that uses unstructured data:", ml_task)
```
